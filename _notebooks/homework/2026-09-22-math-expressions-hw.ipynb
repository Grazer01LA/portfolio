# 3.3 Math Expressions

Practice building Python expressions from variables. Each activity uses arithmetic to calculate a result instead of typing the result directly.

Python follows the usual order of operations: parentheses, exponents, multiplication and division, then addition and subtraction. The `//` operator gives a whole-number quotient, while `%` gives the remainder.

## Popcorn Hack 1: Arithmetic warm-up

Use the three values to calculate a sum, a difference, a product, and a quotient. Try changing the values after running the cell.

```python
first_value = 48
second_value = 6
third_value = 5

sum_result = first_value + second_value + third_value
difference_result = first_value - second_value
product_result = second_value * third_value
quotient_result = first_value / second_value

print("Sum:", sum_result)
print("Difference:", difference_result)
print("Product:", product_result)
print("Quotient:", quotient_result)
```

## Popcorn Hack 2: Organize the supply kits

A club has cartons of supply kits plus a few extra kits. Calculate the total, then find how many full tables of 12 kits can be set up and how many kits will be left over. Use `//` and `%` for the last two calculations.

```python
kits_per_carton = 18
carton_count = 7
extra_kits = 23
kits_per_table = 12

total_kits = kits_per_carton * carton_count + extra_kits
full_tables = total_kits // kits_per_table
leftover_kits = total_kits % kits_per_table

print("Total kits:", total_kits)
print("Full tables:", full_tables)
print("Kits left over:", leftover_kits)
```

## Homework: Estimate the maker fair budget

A maker fair sells tickets for two days. Use variables and expressions to calculate ticket revenue, total expenses, and the amount remaining after expenses. Then calculate the average ticket revenue per day. Keep the arithmetic in expressions so the report updates when the values change.

```python
ticket_price = 6
visitors_per_day = 35
fair_days = 2
supply_cost_per_day = 42
booth_rental = 85

ticket_revenue = ticket_price * visitors_per_day * fair_days
total_expenses = supply_cost_per_day * fair_days + booth_rental
money_after_expenses = ticket_revenue - total_expenses
average_revenue_per_day = ticket_revenue / fair_days

print("Ticket revenue:", ticket_revenue)
print("Total expenses:", total_expenses)
print("Money after expenses:", money_after_expenses)
print("Average ticket revenue per day:", average_revenue_per_day)
```
