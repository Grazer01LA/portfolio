## Popcorn Hack 1: Index a supply list

Python's last index is 3, not 4, for this four-item list. AP pseudocode uses one-based indexing, so its last valid index is 4.


```python
items_needed = ["blankets", "socks", "water", "meals"]
print(items_needed[0])
print(items_needed[3])
print(len(items_needed))
assert (items_needed[0], items_needed[3], len(items_needed)) == ("blankets", "meals", 4)
```

## Popcorn Hack 2: Volunteer hours

Traverse the values to calculate total and average, and guard the empty-list case before dividing.


```python
def summarize_hours(hours):
    if not hours:
        return "No data yet"
    total = 0
    for hour in hours:
        total += hour
    return total, total / len(hours)

volunteer_hours = [7, 6, 8, 5, 7, 9]
print("Total and average:", summarize_hours(volunteer_hours))
assert summarize_hours(volunteer_hours) == (42, 7.0)
assert summarize_hours([]) == "No data yet"
```

## Popcorn Hack 3: Count and filter donation requests

Count each requested item and keep items with at least two votes. The popular list is `socks` and `meals`.


```python
items_requested = ["blankets", "socks", "meals", "socks", "water", "socks", "meals"]
frequencies = {}
for item in items_requested:
    frequencies[item] = frequencies.get(item, 0) + 1
popular_items = [item for item, count in frequencies.items() if count >= 2]
print(frequencies)
print(popular_items)
assert popular_items == ["socks", "meals"]
```

## Homework Task: Donation report

Use the supplied donation and cost lists. Avoid `sum()` for the cost total, create a frequency dictionary, and collect items with at least three donations. The integer average is the required whole-number result.


```python
donations_by_item = ["blankets", "socks", "blankets", "water", "socks", "blankets", "meals", "socks", "blankets", "water", "blankets", "socks"]
costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]
total_cost = 0
for cost in costs:
    total_cost += cost
frequencies = {}
for item in donations_by_item:
    frequencies[item] = frequencies.get(item, 0) + 1
items_with_at_least_three = [item for item, count in frequencies.items() if count >= 3]
print(len(donations_by_item))
print(donations_by_item[-1])
print("Total:", total_cost)
print("Average:", total_cost // len(costs))
print(frequencies)
print(items_with_at_least_three)
assert (len(donations_by_item), donations_by_item[-1], total_cost, total_cost // len(costs)) == (12, "socks", 31, 2)
assert frequencies == {"blankets": 5, "socks": 4, "water": 2, "meals": 1}
assert items_with_at_least_three == ["blankets", "socks"]
```
