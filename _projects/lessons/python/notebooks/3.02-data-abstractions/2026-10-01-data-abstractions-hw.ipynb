## Popcorn Hacks: Find and extend the levels

A list lets related level values stay together. Python index 1 is the second element; AP pseudocode index 2 refers to that same element. Replacing an element preserves the list length, while appending adds another element.


```python
level_stars = [10, 20, 15]
print(level_stars[0], level_stars[1], len(level_stars))
assert (level_stars[0], level_stars[1], len(level_stars)) == (10, 20, 3)
level_stars[0] = 14
print(level_stars[0], level_stars[1], len(level_stars))
assert (level_stars[0], level_stars[1], len(level_stars)) == (14, 20, 3)

sequence = [6, 12, 12]
sequence.append(5)
print(sequence, len(sequence))
assert (sequence, len(sequence)) == ([6, 12, 12, 5], 4)
sequence[3] = 8
assert (sequence, len(sequence)) == ([6, 12, 12, 8], 4)
sequence.append(8)
assert (sequence, len(sequence)) == ([6, 12, 12, 8, 8], 5)
```

## Homework Hack: Grow a UESL game

Add 5 stars to level two, append a level with 7 stars, and traverse the resulting collection to report its count and total. The empty-list test demonstrates that a running-total algorithm safely returns zero without trying to access a missing element. Lists prevent the program from needing a separately named variable for every level.


```python
level_stars = [12, 18, 10]
level_stars[1] += 5
level_stars.append(7)

def report_levels(levels):
    total = 0
    for stars in levels:
        total += stars
    print("Levels:", levels)
    print("Count:", len(levels))
    print("Total stars:", total)
    return total

assert report_levels(level_stars) == 52
assert report_levels([]) == 0

def add_to_second_level(levels, bonus):
    if len(levels) > 1:
        levels[1] += bonus

add_to_second_level(level_stars, 1)
assert level_stars[1] == 24
empty_levels = []
add_to_second_level(empty_levels, 5)
assert empty_levels == []
```
