## Popcorn Hack 1: The silent failure

The original conditional has no false branch, so a zero request prints nothing. That is not a Python error, but it is a logic/UX bug because the user gets no answer. Add an `else` so both outcomes are handled.


```python
from contextlib import redirect_stdout
from io import StringIO

amount_requested = 0
captured = StringIO()
with redirect_stdout(captured):
    if amount_requested > 0:
        print("Request received")
assert captured.getvalue() == ""

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 0:
    fixed_message = "Request received"
else:
    fixed_message = "Enter an amount greater than zero"
print(fixed_message)
assert fixed_message == "Enter an amount greater than zero"
```

## Popcorn Hack 2: Put the most specific branch first

With a score of 9, testing `score >= 5` first would capture the value before the `score >= 8` critical case. Ordering thresholds from highest to lowest keeps each branch reachable.


```python
def classify_risk(score):
    if score >= 8:
        return "Critical"
    elif score >= 5:
        return "Elevated"
    else:
        return "Low"

print(classify_risk(9))
assert [classify_risk(score) for score in (2, 6, 9)] == ["Low", "Elevated", "Critical"]
```

## Homework Hack: Build the trainer's scoring step

Score each independent signal without an `else`: a gift-card phrase adds 4, an amount over 100 adds 3, and an unknown sender adds 2. Then classify the total using a descending, mutually exclusive chain with five reachable outcomes. The three signals total 9; a clean known message under the amount threshold scores 0.


```python
def score_message(message, amount, sender_known):
    score = 0
    if "gift cards" in message.lower():
        score += 4
    if amount > 100:
        score += 3
    if not sender_known:
        score += 2
    return score

def report_risk(score):
    if score >= 9:
        return "Urgent review"
    elif score >= 7:
        return "High risk"
    elif score >= 4:
        return "Review"
    elif score >= 1:
        return "Low risk"
    else:
        return "No risk signals"

all_signals = score_message("Buy gift cards now", 500, False)
clean_score = score_message("Hello, thank you", 50, True)
print("All signals:", all_signals, report_risk(all_signals))
print("Clean message:", clean_score, report_risk(clean_score))
assert all_signals == 9 and clean_score == 0
assert len({report_risk(score) for score in (0, 2, 5, 7, 9)}) == 5
```
