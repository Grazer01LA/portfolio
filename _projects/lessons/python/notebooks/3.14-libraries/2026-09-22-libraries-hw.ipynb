## Popcorn Hack 1: Read the API, then use it

Use the supplied points, change the jersey range to 1–55, and find the square root of the supplied 144-square-foot court. The mean is 24.4 and rounds up to 25; the court side is 12.0 feet. The random jersey varies, but remains in range.


```python
import math
import random
import statistics

points_per_game = [24, 18, 31, 27, 22]
average_points = statistics.mean(points_per_game)
rounded_average = math.ceil(average_points)
jersey_number = random.Random(14).randint(1, 55)
court_side = math.sqrt(144)
print("Average:", average_points, "rounded up:", rounded_average)
print("Jersey:", jersey_number)
print("Court side:", court_side, "feet")
assert (average_points, rounded_average, court_side) == (24.4, 25, 12.0)
assert 1 <= jersey_number <= 55
```

## Popcorn Hack 2: Write and test Flask routes

The score and roster routes return the supplied data. The additional `/coach` route is registered before the test client is created, and all three endpoints are checked with Flask's test client.


```python
from flask import Flask, jsonify

app = Flask(__name__)

@app.get("/score")
def score():
    return jsonify({"home": 58, "away": 52, "quarter": 3})

@app.get("/roster")
def roster():
    return jsonify({"players": ["Maya", "Leo", "Avery"]})

@app.get("/coach")
def coach():
    return jsonify({"coach": "Coach Rivera"})

client = app.test_client()
assert client.get("/score").get_json() == {"home": 58, "away": 52, "quarter": 3}
assert client.get("/roster").get_json() == {"players": ["Maya", "Leo", "Avery"]}
assert client.get("/coach").get_json() == {"coach": "Coach Rivera"}
print("Flask routes passed")
```

## Homework Hack: Season stats report

This report uses the standard-library `datetime`, `statistics`, and `json` modules. It returns total points, average points, best game, and points by opponent, and calculates days until the next game. For a reproducible example, the calculation is evaluated as of 2026-09-22; replace that date with `date.today()` for a live report. Standard-library modules are not listed in requirements because they ship with Python.

`requirements.txt` lists `pandas` and `Flask`, the third-party packages used by the project.


```python
import json
from datetime import date
import statistics

games = [
    {"opponent": "Lincoln", "points": 24},
    {"opponent": "Madison", "points": 18},
    {"opponent": "Lincoln", "points": 31},
    {"opponent": "Roosevelt", "points": 27},
]
next_game = date(2026, 10, 15)

def summarize_season(games):
    """Return total, average, best game, and points grouped by opponent."""
    points = [game["points"] for game in games]
    points_by_opponent = {}
    for game in games:
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]
    best_game = max(games, key=lambda game: game["points"])
    return {
        "total_points": sum(points),
        "average_points": statistics.mean(points),
        "best_game": best_game["points"],
        "points_per_opponent": points_by_opponent,
    }

stats = summarize_season(games)
days_until_next_game = (next_game - date(2026, 10, 1)).days
print(json.dumps(stats, sort_keys=True))
print("Days until next game:", days_until_next_game)
assert stats == {
    "total_points": 100,
    "average_points": 25,
    "best_game": 31,
    "points_per_opponent": {"Lincoln": 55, "Madison": 18, "Roosevelt": 27},
}
assert days_until_next_game == 14
```
