## Mini-game: Name the procedures

A descriptive procedure name says what the sequence accomplishes. The interactive ingredient is selected from the page's choices; each named procedure then performs one clear step.


```python
import random

def choose_ingredient(rng):
    return rng.choice(["flour", "sugar", "eggs", "milk"])

def mix_ingredients(ingredient):
    return f"Mix the {ingredient} into the batter"

def heat_dessert():
    return "Bake until ready"

def make_dessert(rng):
    ingredient = choose_ingredient(rng)
    return [f"Selected {ingredient}", mix_ingredients(ingredient), heat_dessert(), "Dessert is ready"]

dessert_steps = make_dessert(random.Random(7))
for step in dessert_steps:
    print(step)
assert dessert_steps[0].split()[-1] in {"flour", "sugar", "eggs", "milk"}
assert dessert_steps[-1] == "Dessert is ready"
```

## Suggested practice: Rename, translate, and split

Replacing vague names such as `do_step` with names like `choose_ingredient` makes calls self-documenting. In Java, these steps could be `private` methods called by a public `makeDessert` method; access modifiers determine which code can call them. Splitting a long procedure into smaller procedures also lets each behavior be tested independently.


```python
def is_supported_ingredient(ingredient):
    return ingredient in {"flour", "sugar", "eggs", "milk"}

assert is_supported_ingredient("eggs")
assert not is_supported_ingredient("pepper")
assert mix_ingredients("sugar") == "Mix the sugar into the batter"
```
