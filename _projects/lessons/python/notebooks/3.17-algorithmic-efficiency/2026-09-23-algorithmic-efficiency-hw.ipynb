## Popcorn warm-up: Find the book

For the supplied target, linear search takes five checks and binary search takes three, saving two. If the target is the first item, linear search takes one while binary search still takes three; the negative savings show that binary search is not always faster for a particular target.


```python
linear_checks = 5
binary_checks = 3
checks_saved = linear_checks - binary_checks
print(linear_checks, binary_checks, checks_saved)
assert (linear_checks, binary_checks, checks_saved) == (5, 3, 2)

linear_checks = 1
binary_checks = 3
checks_saved = linear_checks - binary_checks
print(linear_checks, binary_checks, checks_saved)
assert checks_saved == -2
```

## Popcorn checkpoint: Expand the catalog

A worst-case binary search adds about one check when the catalog size doubles from 8 to 16, while a linear scan doubles its checks. A logarithmic estimate assumes the catalog is sorted and can be accessed by index.


```python
import math

def binary_worst_case(n):
    return 0 if n == 0 else math.floor(math.log2(n)) + 1

for n in (8, 16):
    print(n, n, binary_worst_case(n))
assert binary_worst_case(8) == 4
assert binary_worst_case(16) == 5
```

## Homework Hack: Advise the librarian

Compare the supplied linear visits and ordered-pair work at 12, 24, and 0 entries. Pair work grows quadratically: doubling the input quadruples that count. For repeated searches of 1,000 sorted entries, binary search needs at most 10 checks instead of 1,000. If the data is unsorted, sorting has an upfront cost; if it must be copied into sorted order, that also uses memory.

A librarian should prefer binary search for many searches once a catalog is sorted. For a single search in unsorted data, the cost of sorting first may not be worthwhile.


```python
def growth_counts(n):
    book_count = n
    visits = n
    ordered_pairs = n * n
    return book_count, visits, ordered_pairs

for n, expected in ((12, (12, 12, 144)), (24, (24, 24, 576)), (0, (0, 0, 0))):
    result = growth_counts(n)
    print(*result)
    assert result == expected

linear_for_1000 = 1000
binary_for_1000 = binary_worst_case(1000)
print("For 1,000 sorted entries:", linear_for_1000, "linear checks versus", binary_for_1000, "binary checks")
assert binary_for_1000 == 10
```
