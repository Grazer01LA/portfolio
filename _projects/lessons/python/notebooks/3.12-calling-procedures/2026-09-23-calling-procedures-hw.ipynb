## Popcorn Hack 1: Simple procedure calls and parameters

The same procedure can report different values when called with different arguments.


```python
def report_status(location, status):
    message = f"Status update from {location}: {status}"
    print(message)
    return message

assert report_status("Poway", "All clear") == "Status update from Poway: All clear"
assert report_status("Ramona", "Road closed") == "Status update from Ramona: Road closed"
```

## Popcorn Hack 2: Decomposition and helper procedures

A helper classifies an incident; the larger procedure calls it to choose a message. The two supplied calls produce the priority and standard report lines.


```python
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

def process_report(location, incident_type):
    if is_priority(incident_type):
        message = f"PRIORITY: {incident_type} reported in {location}"
    else:
        message = f"STANDARD REPORT: {incident_type} reported in {location}"
    print(message)
    return message

assert process_report("Poway", "Medical Emergency") == "PRIORITY: Medical Emergency reported in Poway"
assert process_report("Poway", "Power Outage") == "STANDARD REPORT: Power Outage reported in Poway"
```

## Homework Hack: Dispatch response

The dispatch procedure uses the Boolean returned by a separate priority-check helper. Three calls cover different incident types and two locations.


```python
def check_priority(incident_type):
    return incident_type in {"Medical Emergency", "Fire"}

def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        return f"Send an immediate response to {location}: {incident_type}"
    return f"Queue a standard response for {location}: {incident_type}"

responses = [
    dispatch_response("Poway", "Medical Emergency"),
    dispatch_response("Ramona", "Power Outage"),
    dispatch_response("Poway", "Fire"),
]
for response in responses:
    print(response)
assert len(responses) == 3
assert responses[0].startswith("Send an immediate")
assert responses[1].startswith("Queue a standard")
```
