## Popcorn Hack: Checkpoint challenge

A checkpoint stores the current value; changing the live variable afterward does not change that saved value. Starting with 8 stars, the live value becomes 12 while the checkpoint stays 8.


```python
stars = 8
checkpoint_stars = stars
stars += 4
print("Live stars:", stars)
print("Checkpoint stars:", checkpoint_stars)
assert (stars, checkpoint_stars) == (12, 8)
```

## Homework Hack: Build a UESL star counter

Keep each level's stars in a separate variable, report the game and reduced-motion choice, then compare the stored total with a total recalculated after a level-two bonus. The bonus is tested with both 5 and 0. Reduced motion should be a player choice because animation can be uncomfortable or inaccessible for some players.


```python
game_name = "UESL"
reduced_motion = True
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

def report_stars(bonus):
    stored_total = level_one_stars + level_two_stars + level_three_stars
    adjusted_level_two = level_two_stars + bonus
    recalculated_total = level_one_stars + adjusted_level_two + level_three_stars
    print(f"Game: {game_name}; reduced motion: {reduced_motion}")
    print("Stored total:", stored_total)
    print("Recalculated total:", recalculated_total)
    return stored_total, recalculated_total

assert report_stars(5) == (40, 45)
assert report_stars(0) == (40, 40)
```
