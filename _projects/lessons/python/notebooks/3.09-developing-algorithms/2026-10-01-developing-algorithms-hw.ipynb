## Popcorn Hack 1: Translate pseudocode to Python

Use iteration to calculate the total and average without calling `sum()`.


```python
hours = [3, 5, 2, 4]
total = 0
for hour in hours:
    total += hour
average = total / len(hours)
print(total)
print(average)
assert (total, average) == (14, 3.5)
```

## Popcorn Hack 2: Predict, then fix

The original strict `>` misses the workshop exactly at capacity. Changing it to `>=` counts both 12 and 15 as full.


```python
signed_up = [12, 7, 15, 9]
capacity = 12
full = 0
for count in signed_up:
    if count >= capacity:
        full += 1
print("Full workshops:", full)
assert full == 2
```

## Popcorn Hack 3: Modify a counter into a collector

Collect the names of workshops with open seats rather than only counting them.


```python
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
open_workshops = []
for index in range(len(workshops)):
    if signed_up[index] < 12:
        open_workshops.append(workshops[index])
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
assert open_workshops == ["Parenting Skills", "Self-Sufficiency"]
```

## Homework Hack: Supply closet restock report

Compare in-stock quantities with targets, collect each shortage, total shortages with a loop, and find the largest shortage. Items at the target amount are not short.


```python
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]
shortages = []
for index in range(len(items)):
    shortage = target[index] - in_stock[index]
    if shortage > 0:
        shortages.append((items[index], shortage))

total_needed = 0
for item, amount in shortages:
    total_needed += amount

if shortages:
    most_urgent = shortages[0]
    for shortage in shortages[1:]:
        if shortage[1] > most_urgent[1]:
            most_urgent = shortage
    print("Restock needed:")
    for item, amount in shortages:
        print(f"{item}: need {amount}")
    print("Total items needed:", total_needed)
    print(f"Most urgent: {most_urgent[0]} (short by {most_urgent[1]})")
else:
    print("Fully stocked!")
assert shortages == [("Blankets", 12), ("Shampoo", 10)]
assert total_needed == 22 and most_urgent == ("Blankets", 12)
```
