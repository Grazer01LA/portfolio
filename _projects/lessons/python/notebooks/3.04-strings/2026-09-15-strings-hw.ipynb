## Popcorn Hack 1: Extract a hidden message

Locate the tokens with string indexing and slicing, print the length, and inspect every other character.


```python
signal = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"
signal_start = signal.index("SIGNAL")
confirmed_start = signal.index("CONFIRMED")
delta_start = signal.index("DELTA")
print(signal[signal_start:signal_start + 6])
print(signal[confirmed_start:confirmed_start + 9])
print(signal[-5:])
print(len(signal))
print(signal[::2])
assert (signal[signal_start:signal_start + 6], signal[confirmed_start:confirmed_start + 9], signal[-5:], len(signal)) == ("SIGNAL", "CONFIRMED", "DELTA", 31)
```

## Popcorn Hack 2: Scrub and search a noisy transmission

Remove surrounding whitespace and transmission noise, search without regard to case, and count the cleaned words.


```python
raw = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "
cleaned = raw.strip()
for noise in (">", "-", "!"):
    cleaned = cleaned.replace(noise, "")
lowercase = cleaned.lower()
words = cleaned.split()
print(cleaned)
print("arrived" in lowercase)
print("Word count:", len(words))
assert "arrived" in lowercase
assert len(words) == 5
```

## Popcorn Hack 3: Build a decoded report

Concatenate the headline, format a numeric confidence, and join message fragments into one report.


```python
agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87
fragments = ["ALPHA", "SEVEN", "DELTA"]
headline = agent_name + " reports target " + target_word
confidence_line = "Confidence: " + str(confidence) + "%"
confidence_f_string = f"Confidence: {confidence}%"
fragment_line = "-".join(fragments)
report = f"{headline}\n{confidence_line}\nFragments: {fragment_line}"
print(report)
assert confidence_line == confidence_f_string
assert fragment_line == "ALPHA-SEVEN-DELTA"
```

## Homework Hack: Crack the final transmission

Parse the CSV fields, inspect the year with a slice, decode the dash-separated message, and prepare Markdown and JSON-style records. JSON Boolean values are lowercase.


```python
row = "2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87"
date, agent, message_code, confidence_text = row.strip().split(",")
year_is_2026 = date[:4] == "2026"
message = message_code.replace("-", " ")
contains_dawn = "dawn" in message.lower()
markdown_row = f"| {date} | {agent} | {message} | {confidence_text}% |"
json_line = f'{{"date":"{date}","agent":"{agent}","message":"{message}","confidence":{confidence_text},"year_is_2026":{str(year_is_2026).lower()},"contains_dawn":{str(contains_dawn).lower()}}}'
print(markdown_row)
print(json_line)
assert year_is_2026 and contains_dawn
assert message == "THE EAGLE LANDS AT DAWN"
```
