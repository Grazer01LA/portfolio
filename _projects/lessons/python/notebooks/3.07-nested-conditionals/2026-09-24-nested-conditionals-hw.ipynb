## Popcorn Hack 1: After-school homework plan

The weekday decision is the outer branch. Only on a weekday do we check whether homework is done; on a weekend, that inner check is not needed.


```python
def after_school_plan(is_weekday, homework_finished):
    if is_weekday:
        if homework_finished:
            return "Free time"
        else:
            return "Finish homework"
    else:
        return "Weekend plan"

print(after_school_plan(True, False))
assert after_school_plan(True, False) == "Finish homework"
assert after_school_plan(True, True) == "Free time"
assert after_school_plan(False, False) == "Weekend plan"
```

## Popcorn Hack 2: Repair the student account check

Check login first. Permission is meaningful only after a user has logged in, so it belongs inside the login branch.


```python
def account_message(is_logged_in, has_permission):
    if is_logged_in:
        print("Account found")
        if has_permission:
            return "Page opened"
        else:
            return "Permission needed"
    else:
        return "Log in first"

assert [account_message(*case) for case in ((False, False), (True, False), (True, True))] == [
    "Log in first", "Permission needed", "Page opened"
]
print(account_message(False, False))
print(account_message(True, False))
print(account_message(True, True))
```

## Popcorn Hack 3: Lunch and sports practice

The four paths are tested below. The same logic can be translated directly: JavaScript uses `if (...) { ... } else { ... }`, and College Board pseudocode uses `IF (condition) { ... } ELSE { ... }`. The outer lunch check is evaluated first; the practice check is evaluated inside each outcome.


```python
def lunch_and_practice(packed_lunch, has_practice):
    if packed_lunch:
        if has_practice:
            return "Eat your packed lunch, then go to practice"
        else:
            return "Eat your packed lunch, then enjoy free time"
    else:
        if has_practice:
            return "Get lunch at school, then go to practice"
        else:
            return "Get lunch at school, then enjoy free time"

outcomes = {lunch_and_practice(lunch, practice) for lunch in (False, True) for practice in (False, True)}
assert len(outcomes) == 4
print("Four lunch/practice paths tested:", len(outcomes))
```

## Homework Hack: High School Daily Routine Dashboard

Login is checked before any schedule details. A logged-in student receives decisions for homework, lunch, and practice, yielding at least four recommendations. The tests cover both login outcomes and all eight combinations of the three schedule values.


```python
def routine_dashboard(logged_in, homework_done, packed_lunch, practice):
    if not logged_in:
        return ["Log in first to view your routine"]
    recommendations = []
    if homework_done:
        recommendations.append("Homework is complete; plan a break")
    else:
        recommendations.append("Set aside time to finish homework")
    if packed_lunch:
        recommendations.append("Remember to bring your lunch")
    else:
        recommendations.append("Plan to get lunch at school")
    if practice:
        recommendations.append("Pack sports gear for practice")
    else:
        recommendations.append("Use the afternoon for another activity")
    return recommendations

student = {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True}
starter_recommendations = routine_dashboard(student["logged_in"], student["homework_finished"], student["lunch_packed"], student["practice_today"])
print("Starter recommendations:", starter_recommendations)
assert routine_dashboard(False, True, True, True) == ["Log in first to view your routine"]
for homework_done in (False, True):
    for packed_lunch in (False, True):
        for practice in (False, True):
            assert len(routine_dashboard(True, homework_done, packed_lunch, practice)) == 3
print("All 9 dashboard outcomes passed")
```
