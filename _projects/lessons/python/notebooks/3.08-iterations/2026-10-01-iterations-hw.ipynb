## Popcorn Hack: Wildfire dispatch scanner

Safe alerts are skipped with `continue`; every action alert is counted and totaled; the loop stops at the first critical alert with `break`.


```python
alerts = [1, 4, 2, 6, 3, 9, 5]
safe_count = 0
action_count = 0
action_total = 0
critical_count = 0
for alert in alerts:
    if alert < 3:
        safe_count += 1
        continue
    action_count += 1
    action_total += alert
    if alert >= 8:
        critical_count += 1
        print("Dispatch immediately:", alert)
        break
print("Summary:", safe_count, action_count, action_total, critical_count)
assert (safe_count, action_count, action_total, critical_count) == (2, 4, 22, 1)
```

## Homework Hack: Loop-based air-quality checker

This monitoring example uses eight readings, a safe threshold below 20, a warning band, and a critical threshold at 70. Safe readings continue without warning work; the first critical reading triggers `break`. Three counts plus a warning total form the final report.


```python
readings = [15, 25, 60, 18, 42, 80, 30, 55]
safe_readings = 0
warning_readings = 0
critical_readings = 0
warning_total = 0
for reading in readings:
    if reading < 20:
        safe_readings += 1
        continue
    if reading >= 70:
        critical_readings += 1
        break
    warning_readings += 1
    warning_total += reading
print("Air-quality summary:", safe_readings, warning_readings, critical_readings, warning_total)
assert (safe_readings, warning_readings, critical_readings, warning_total) == (2, 3, 1, 127)
```
