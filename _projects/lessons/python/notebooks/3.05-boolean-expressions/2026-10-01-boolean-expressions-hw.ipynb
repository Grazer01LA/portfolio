## Popcorn Hack 1: Check one SFI part record

Validate the four required fields independently. A record must use an accepted racing category.


```python
accepted_categories = {"Auto Racing", "Drag Racing"}
part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001",
}
checks = (
    bool(part["product_name"]),
    part["category"] in accepted_categories,
    bool(part["spec_number"]),
    bool(part["effective_date"]),
)
print("Required field checks:", checks)
assert checks == (True, True, True, True)
changed_part = {**part, "category": "Street Car"}
assert changed_part["category"] not in accepted_categories
```

## Popcorn Hack 2: Backend create rule

A record can be created only when every required field passes and its spec number is not already stored. The supplied 1.1 record is rejected as a duplicate; a valid new 3.1 spec is accepted.


```python
existing_spec_numbers = ["1.1", "2.1"]

def can_create(record, existing_specs):
    fields_valid = (
        bool(record.get("product_name"))
        and record.get("category") in accepted_categories
        and bool(record.get("spec_number"))
        and bool(record.get("effective_date"))
    )
    return fields_valid and record["spec_number"] not in existing_specs

print("Duplicate allowed:", can_create(part, existing_spec_numbers))
new_part = {**part, "spec_number": "3.1"}
print("New valid record allowed:", can_create(new_part, existing_spec_numbers))
assert can_create(part, existing_spec_numbers) is False
assert can_create(new_part, existing_spec_numbers) is True
```

## Popcorn Hack 3: SFI search filter

Search by product-name substring or exact spec number while restricting results to supported racing categories. Query `1.2` returns only the matching clutch assembly.


```python
records = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"},
]

def search_parts(query, catalog):
    query_lower = query.lower()
    return [record for record in catalog
            if record["category"] in accepted_categories
            and (query_lower in record["product_name"].lower() or query == record["spec_number"])]

matches = search_parts("1.2", records)
print("Matches:", matches)
assert [record["spec_number"] for record in matches] == ["1.2"]
assert search_parts("not found", records) == []
```

## Homework Hack: SFI backend validator

The validator checks nonempty name, supported category, nonempty spec, effective date, and uniqueness. The three supplied candidates should be accepted, rejected for category, and rejected as a duplicate, respectively.


```python
candidates = [
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.2", "effective_date": "Feb. 9, 2006"},
    {"product_name": "Replacement Flywheels", "category": "Street Car", "spec_number": "3.1", "effective_date": "Jan. 1, 2026"},
    {"product_name": "Existing Flywheel Record", "category": "Drag Racing", "spec_number": "1.1", "effective_date": "Jan. 1, 2026"},
]
decisions = [can_create(candidate, existing_spec_numbers) for candidate in candidates]
for candidate, accepted in zip(candidates, decisions):
    print(candidate["product_name"], "accepted" if accepted else "rejected")
assert decisions == [True, False, False]
```
