## 1. Reference Guide

### Key terms

| Term | Meaning | Library example |
| --- | --- | --- |
| Input size | Amount of data, called n | Number of catalog entries |
| Correctness | Producing the right answer | Find the right book or correctly report it missing |
| Time efficiency | How the work grows with input | Count inspected entries |
| Space efficiency | How memory use grows | A second catalog copy needs extra storage |
| Best / worst case | Least / most work at a given size | First match / missing book in a linear search |
| Heuristic | Useful approach without a guarantee of the best result | Choose short books first for a reading-time budget |

### Growth patterns

| Growth | Pattern | Example |
| --- | --- | --- |
| Constant | Stays the same | Read one known catalog position |
| Logarithmic | Grows slowly through repeated halving | Worst-case binary search |
| Linear | Doubling n doubles work | Worst-case linear search |
| Quadratic | Doubling n gives four times the work | Compare every book with every book |
| Exponential | Each additional item can multiply choices | Consider every possible selection of books: 2^n choices |


```python
linear_checks = 5
binary_checks = 3
print(linear_checks)
print(binary_checks)
print(linear_checks - binary_checks)
```

### Warm-up reflection

Linear search inspects entries one by one. Binary search repeatedly halves a sorted catalog. Both can be correct, but the work is different.

Why can two correct methods produce different counts of checks?


```python
import math

n = 8
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)
```

### Compare growth

For a sorted catalog, binary search is often much faster than a linear scan as the catalog grows. But binary search requires sorted data and direct access to the middle of the list.

The next example compares a scan with a pair-counting task.


```python
n = 10
visits = n
pairs = n * n
print(n)
print(visits)
print(pairs)
```

## 4. Summary

Algorithmic efficiency is about comparing correct methods by the amount of work they require as the input grows. A linear search may be simple and correct, but a binary search can be far more efficient when the data is sorted and direct access is available.

Comparing growth patterns helps us make informed design decisions in real-world problems like library catalogs and information retrieval.
