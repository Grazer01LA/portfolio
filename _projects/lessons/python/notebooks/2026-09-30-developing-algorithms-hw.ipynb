## 1. Reference Guide

### Key Topics

| Topic | Meaning | Example |
| --- | --- | --- |
| Algorithm | A finite set of instructions | Add donations and check the goal |
| Sequencing | Running steps in order | Set total = 0, then add, then print |
| Selection | Choosing what to do with an if | if total >= goal: |
| Iteration | Repeating steps with a loop | for amount in donations: |
| Equivalent algorithms | Same result and same side effects | Two seat checks always agree |
| Side effect | A change besides the returned result | .sort() reorders the original list |
| Boundary input | A value exactly at a limit | 12 of 12 seats filled |
| Modify / Combine | Change a known algorithm or use two together | Counter to collector |

### Python Algorithm Tools

| Tool | Pattern | Example |
| --- | --- | --- |
| Counter | Counts how many items match | count = count + 1 |
| Running total | Adds values as you go | total = total + amount |
| Collector | Builds a list of matching items | names.append(name) |
| Max finder | Tracks the largest value seen | if x > largest: largest = x |
| Comparison | Decides pass or fail at a limit | if signed_up >= 12: |
| Function | Packages an algorithm to reuse and test | def has_seat(signed_up): |


### College Board pseudocode connections

| Python | Pseudocode | Meaning |
| --- | --- | --- |
| total = 0 | total <- 0 | Store or update a value |
| print(total) | DISPLAY(total) | Show a result |
| if condition: | IF (condition) { } | Choose between paths |
| for item in list: | FOR EACH item IN list { } | Visit every item |
| list.append(value) | APPEND(list, value) | Add to the end |
| len(list) | LENGTH(list) | Count items |
| list[0] | list[1] | Python indexes from 0, pseudocode from 1 |


````python
donations = [25, 150, 40, 100, 10, 250]
goal = 500
total = 0
for amount in donations:
    total = total + amount
if total >= goal:
    print(f"Goal met! Safe Nights raised ${total}.")
else:
    print(f"Still need ${goal - total} to reach the goal.")
````

total <- 0

## 2. Same Result, or Not?

Two algorithms can look different and still be equivalent. Two can look almost identical and do different things.

Boundary values matter. Test below, at, and above the limit.


````python
donations = [25, 150, 40, 100, 10, 250]
count_a = 0
for amount in donations:
    if amount >= 100:
        count_a = count_a + 1
print("Algorithm A counted:", count_a)

count_b = 0
for amount in donations:
    if amount >= 100:
        count_b = count_b + 1
    count_b = count_b + 1
print("Algorithm B counted:", count_b)
````

## 3. Modify and Combine

Most new algorithms are built by modifying a known pattern or combining two patterns.

A counter becomes a collector when you replace `count = count + 1` with an append.

### Popcorn Hack 3: Modify the counter into a collector (2 min)

Staff want the names of workshops that still have a seat (fewer than 12). The expected output is:

```text
Workshops with open seats: 2
['Parenting Skills', 'Self-Sufficiency']
```


````python
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
capacity = 12
open_workshops = []
for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
````

## 4. Homework Hack: Supply Closet Restock Report

Safe Passage Heals keeps a supply closet with target amounts. Before a donation drive, staff need a restock report.

Expected output:

```text
Restock needed:
Blankets: need 12
Shampoo: need 10
Total items needed: 22
Most urgent: Blankets (short by 12)
```


````python
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]
short_items = []
short_amounts = []
for i in range(len(items)):
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])

total_needed = 0
for amount in short_amounts:
    total_needed = total_needed + amount

max_index = 0
for i in range(1, len(short_amounts)):
    if short_amounts[i] > short_amounts[max_index]:
        max_index = i

if short_items:
    print("Restock needed:")
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
    print(f"Total items needed: {total_needed}")
    print(f"Most urgent: {short_items[max_index]} (short by {short_amounts[max_index]})")
else:
    print("Fully stocked!")
````
