## Extension: Find a shortest path through the maze

Use breadth-first search to explore open cells, remember each cell's predecessor, and reconstruct a shortest route from start to goal. `0` is open and `1` is a wall.


```python
from collections import deque

maze = [
    [0, 0, 1, 0],
    [1, 0, 1, 0],
    [0, 0, 0, 0],
    [0, 1, 1, 0],
]
start = (0, 0)
goal = (3, 3)

def shortest_path(grid, start, goal):
    queue = deque([start])
    previous = {start: None}
    while queue:
        row, column = queue.popleft()
        if (row, column) == goal:
            break
        for next_row, next_column in ((row - 1, column), (row + 1, column), (row, column - 1), (row, column + 1)):
            next_cell = (next_row, next_column)
            inside = 0 <= next_row < len(grid) and 0 <= next_column < len(grid[0])
            if inside and grid[next_row][next_column] == 0 and next_cell not in previous:
                previous[next_cell] = (row, column)
                queue.append(next_cell)
    if goal not in previous:
        return []
    path = []
    cell = goal
    while cell is not None:
        path.append(cell)
        cell = previous[cell]
    return list(reversed(path))

path = shortest_path(maze, start, goal)
print("Path:", path)
assert path[0] == start and path[-1] == goal
assert all(maze[row][column] == 0 for row, column in path)
assert all(abs(r1 - r2) + abs(c1 - c2) == 1 for (r1, c1), (r2, c2) in zip(path, path[1:]))
```
