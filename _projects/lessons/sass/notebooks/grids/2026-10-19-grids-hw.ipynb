## Popcorn Hack 2: Size the Beakers with `fr`

Use four `fr` values for four columns, with the Control Beaker taking twice the share of each other beaker. The completed rule is `grid-template-columns: 1fr 1fr 1fr 2fr;`.


```python
%%html
<!-- UI_RUNNER: Grid Popcorn Hack 2 -->
<style>
  #beaker-fr-grid {
    display: grid;
    grid-template-columns: 1fr 1fr 1fr 2fr;
    gap: 0.5rem;
  }
</style>
<div id="beaker-fr-grid">
  <div>Beaker 1</div>
  <div>Beaker 2</div>
  <div>Beaker 3</div>
  <div>Control Beaker</div>
</div>
```

## Homework Hack: Beaker Report and Calculator

Complete both required grids using OCS classes only. The standard grid has a header and an accent cell for the Beaker 3 outlier; the calculator grid accents its Avg key. The checklist reports PASS for each requirement.


```python
%%html
<!-- UI_RUNNER: Grid Homework -->
<p id="grid-checklist" aria-live="polite"></p>
<div id="reading-grid" class="ocs__grid ocs__grid--standard cols-4">
  <div class="ocs__grid-cell ocs__grid-cell--header">Trial 4: Beaker Readings</div>
  <div class="ocs__grid-cell">Beaker 1: 18 mL</div>
  <div class="ocs__grid-cell">Beaker 2: 19 mL</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 34 mL (outlier)</div>
  <div class="ocs__grid-cell">Control Beaker: 18 mL</div>
</div>
<div id="calculator-grid" class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell">7</div>
  <div class="ocs__grid-cell">8</div>
  <div class="ocs__grid-cell">9</div>
  <div class="ocs__grid-cell">4</div>
  <div class="ocs__grid-cell">5</div>
  <div class="ocs__grid-cell">6</div>
  <div class="ocs__grid-cell">1</div>
  <div class="ocs__grid-cell">2</div>
  <div class="ocs__grid-cell">3</div>
  <div class="ocs__grid-cell">0</div>
  <div class="ocs__grid-cell">.</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
</div>
<script>
  const readingGrid = document.getElementById("reading-grid");
  const calculatorGrid = document.getElementById("calculator-grid");
  const checks = [
    ["Both grids use one OCS variant", readingGrid.classList.contains("ocs__grid--standard") && calculatorGrid.classList.contains("ocs__grid--calculator")],
    ["Every box uses ocs__grid-cell", [...readingGrid.children, ...calculatorGrid.children].every((cell) => cell.className.includes("ocs__grid-cell"))],
    ["Reading title uses --header and outlier uses --accent", readingGrid.querySelector(".ocs__grid-cell--header") && readingGrid.querySelector(".ocs__grid-cell--accent")],
    ["Calculator uses calculator variant and accented Avg key", calculatorGrid.classList.contains("ocs__grid--calculator") && calculatorGrid.lastElementChild.textContent.trim() === "Avg" && calculatorGrid.lastElementChild.classList.contains("ocs__grid-cell--accent")],
    ["No inline styles or made-up classes", !document.querySelector("#reading-grid [style], #calculator-grid [style]") && [...document.querySelectorAll("#reading-grid [class], #calculator-grid [class]")].every((element) => [...element.classList].every((name) => ["ocs__grid", "ocs__grid--standard", "ocs__grid--calculator", "cols-4", "ocs__grid-cell", "ocs__grid-cell--header", "ocs__grid-cell--accent"].includes(name)))]
  ];
  document.getElementById("grid-checklist").innerHTML = checks.map(([label, passed]) => `${passed ? "PASS" : "FIX"}: ${label}`).join("<br>");
</script>
```
