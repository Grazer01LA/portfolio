## Popcorn Hack 2: Size the Beakers with `fr`

Four `fr` values create four columns. The Control Beaker is twice as wide as each of the other three, so the runner's column rule is `grid-template-columns: 1fr 1fr 1fr 2fr;`.


```python
%%html
<!-- UI_RUNNER: Grid Popcorn Hack 2 -->
<div class="ocs__grid ocs__grid--standard cols-4">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell ocs__grid-cell--wide">Control Beaker</div>
</div>
```

## MCQ Check

1. How many columns does `cols-4` request on a standard grid? **Answer: Four.**
2. Which class marks the grid itself? **Answer: `ocs__grid`.**
3. Which cell type is right for an outlier? **Answer: `ocs__grid-cell--accent`.**
4. What do four values in `1fr 1fr 1fr 2fr` indicate? **Answer: Four columns; the last receives twice the share of each other column.**
