## Popcorn Hacks 1-3: Base, Modifiers, and Action Bar

Each hack is completed in its own runner: the base button class, modifiers with an `ocs__links` group, and status tones in a wide action bar.


```python
%%html
<!-- UI_RUNNER: Button Popcorn Hack 1 -->
<a class="ocs__btn" href="/resources">Resource Directory</a>
```

## Popcorn Hacks 2-3: Modifiers and Wide Action Bar

Add the pill and fill modifiers to the check-in button, group both buttons in `ocs__links`, then use the green status tone and wide action bar for the final example.


```python
%%html
<!-- UI_RUNNER: Button Popcorn Hack 2 -->
<div class="ocs__links">
  <button class="ocs__btn pill accent fill">Complete Daily Check-In</button>
  <button class="ocs__btn outline">Save Draft Journal</button>
</div>
```

```python
%%html
<!-- UI_RUNNER: Button Popcorn Hack 3 -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green fill">Complete A Mental Health Check-In Form</button>
  <a class="ocs__btn alert-green outline" href="/resources">Request Break</a>
  <a class="ocs__btn alert-red fill" href="/resources">Get Crisis Support</a>
</div>
```
