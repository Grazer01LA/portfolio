## Popcorn Hacks 1-3: Base, Modifiers, and Action Bar

The first action receives the base class, the grouped check-in gets pill/fill modifiers inside `ocs__links`, and the final toolbar uses semantic status tones with the wide layout container.


```python
%%html
<!-- UI_RUNNER: Button Popcorn Hacks 1-3 -->
<a class="ocs__btn" href="/resources">Resource Directory</a>
<div class="ocs__links">
  <button class="ocs__btn pill accent fill">Complete Daily Check-In</button>
  <button class="ocs__btn outline">Save Draft Journal</button>
</div>
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green fill">Complete Check-In Form</button>
  <a class="ocs__btn alert-green outline" href="/resources">Request a Break</a>
  <a class="ocs__btn alert-red fill" href="/resources">Get Crisis Support</a>
</div>
```

## Homework Hacks: Resource Toolbar and Action Priority

Both action groups are wrapped in `ocs__links`. Urgent Support is red, Self-Care is green, Community is an accent outline, and the check-in panel uses fill/outline treatment to communicate action priority. No inline styles or custom classes remain.


```python
%%html
<!-- UI_RUNNER: Button Homework Hacks -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="/resources">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="/resources">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="/resources">Community</a>
</div>
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a class="ocs__btn outline" href="/resources">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="/resources">Crisis Hotline</a>
</div>
```

3. Which class groups action controls with OCS spacing? **Answer: `ocs__links`.**
4. Which status tone best signals urgent crisis support? **Answer: `alert-red`.**
