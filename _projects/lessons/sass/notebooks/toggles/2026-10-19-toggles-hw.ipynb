## Popcorn Hack: Toggle Vocabulary and Missing Class

Answers: checkbox, slider, alignment, `border-radius`, white circle. The decorative track needs `ocs__toggle-track` and should be hidden from assistive technology with `aria-hidden="true"`.


```python
%%html
<!-- UI_RUNNER: Toggle Popcorn Hack -->
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input" type="checkbox" id="show-message" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show message</span>
</label>
<p id="message">The checkbox stores whether this message is shown.</p>
<script>
  const showMessage = document.getElementById("show-message");
  const message = document.getElementById("message");
  showMessage.addEventListener("change", () => {
    message.hidden = !showMessage.checked;
  });
</script>
```

## Homework Hack: Three Interactive Settings

Each toggle has a distinct label and purpose. JavaScript updates the setting summary and the live count when any switch changes; comments explain the checkbox state and decorative track. The initial state has two toggles on.


```python
%%html
<!-- UI_RUNNER: Toggle Homework -->
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input setting-toggle" type="checkbox" id="notifications" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Notifications</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input setting-toggle" type="checkbox" id="dark-mode" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Dark mode</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input setting-toggle" type="checkbox" id="focus-mode">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Focus mode</span>
</label>
<p id="settings-summary" aria-live="polite"></p>
<p>Enabled settings: <span id="enabled-count" aria-live="polite">2</span></p>
<script>
  const settingToggles = Array.from(document.querySelectorAll(".setting-toggle"));
  const settingsSummary = document.getElementById("settings-summary");
  const enabledCount = document.getElementById("enabled-count");

  function updateSettings() {
    // The checked property tells us which settings are currently enabled.
    const enabled = settingToggles.filter((toggle) => toggle.checked);
    enabledCount.textContent = String(enabled.length);
    settingsSummary.textContent = enabled.length
      ? `On: ${enabled.map((toggle) => toggle.id.replace("-", " ")).join(", ")}`
      : "All settings are off.";
  }

  // The track is decorative; the associated checkbox remains the control.
  settingToggles.forEach((toggle) => toggle.addEventListener("change", updateSettings));
  updateSettings();
</script>
```
