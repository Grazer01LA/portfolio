## Popcorn Hack: Toggle Vocabulary and Find the Bug

Fill-in answers: checkbox, slider, alignment, `border-radius`, white circle. In the bug challenge, the visible track needs `ocs__toggle-track`; it is decorative, so keep `aria-hidden="true"` on it.


```python
%%html
<!-- UI_RUNNER: Toggle Popcorn Hack -->
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input" type="checkbox" id="show-message" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show message</span>
</label>
<p id="message">The checkbox stores whether this message is shown.</p>
<script>
  const showMessage = document.getElementById("show-message");
  const message = document.getElementById("message");
  showMessage.addEventListener("change", () => {
    message.hidden = !showMessage.checked;
  });
</script>
```

## Homework Hack: Three Interactive Toggles

Each toggle has its own label and purpose, updates a feature on the page, and changes the enabled count. Comments explain the checkbox state and decorative track.


```python
%%html
<!-- UI_RUNNER: Toggle Homework -->
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input feature-toggle" type="checkbox" id="announcements-toggle" data-target="announcements" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show announcements</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input feature-toggle" type="checkbox" id="study-tip-toggle" data-target="study-tip" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show study tip</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input feature-toggle" type="checkbox" id="reminder-toggle" data-target="daily-reminder">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show daily reminder</span>
</label>
<p id="announcements">Today: bring your notebook to class.</p>
<p id="study-tip">Study tip: test one small change at a time.</p>
<p id="daily-reminder" hidden>Remember to save your work before you leave.</p>
<p>Enabled toggles: <span id="enabled-count" aria-live="polite">2</span></p>
<script>
  const featureToggles = Array.from(document.querySelectorAll(".feature-toggle"));
  const enabledCount = document.getElementById("enabled-count");

  function updateFeature(toggle) {
    // The checked property is the checkbox's current on/off state.
    document.getElementById(toggle.dataset.target).hidden = !toggle.checked;
    enabledCount.textContent = String(featureToggles.filter((item) => item.checked).length);
  }

  // The track is decorative, so the labeled checkbox handles interaction.
  featureToggles.forEach((toggle) => {
    toggle.addEventListener("change", () => updateFeature(toggle));
    updateFeature(toggle);
  });
</script>
```
