## Homework Hack: Refactor the Cat Guide

The completed page uses heading hierarchy, paragraphs, an ordered list for the steps, semantic emphasis, and OCS card/section-title classes. It contains no inline styles, style block, or made-up classes.


```python
%%html
<!-- UI_RUNNER: Typography Homework -->
<div class="ocs_card">
  <h2>How to Pat a Cat</h2>
  <h3 class="ocs_section-title">Have you ever wanted to...</h3>
  <p>Pat a cat? Here are a few gentle steps to get started.</p>
  <h3 class="ocs_section-title">How to pat a cat</h3>
  <ol>
    <li>Offer your hand and let the cat approach.</li>
    <li>Gently touch the cat if it seems comfortable.</li>
    <li>Move your hand slowly in the direction of the fur.</li>
  </ol>
  <p><strong>Be gentle</strong> and stop if the cat moves away.</p>
  <h3>Congratulations on a calm greeting!</h3>
</div>
```

## MCQ Check

1. Which element should introduce the page's primary title? **Answer: `<h1>`.**
2. Which element correctly marks three numbered steps? **Answer: `<ol>` containing three `<li>` elements.**
3. What should replace a span used only to make important text bold? **Answer: `<strong>`.**
4. Why use `ocs_lead` instead of inline font styles? **Answer: It uses the shared OCS theme and avoids custom CSS.**
