## Homework Hack: Refactor the Cat Guide

Refactor the lesson's original cat guide: use semantic headings, paragraphs, one ordered list with three steps, `strong` for important text, and the OCS `ocs_card` / `ocs_section-title` classes. Do not add CSS or custom classes.


```python
%%html
<!-- UI_RUNNER: Typography Homework -->
<div class="ocs_card">
  <h2>How to Pat a Cat</h2>
  <h3 class="ocs_section-title">Have you ever wanted to...</h3>
  <p>Pat a cat? Well good for you, we will give step-by-step instructions on how to pat a cat!</p>
  <h3 class="ocs_section-title"><strong>How to pat a cat:</strong></h3>
  <ol>
    <li>Put out your hand.</li>
    <li>Touch the cat.</li>
    <li>Move your hand back and forth.</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```
